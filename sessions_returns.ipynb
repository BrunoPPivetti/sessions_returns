```python
import pandas as pd 
import numpy as np
from zoneinfo import ZoneInfo
```

```python
eurusd = pd.read_csv('EURUSD_H1.csv', sep='\t', usecols=['<DATE>','<TIME>','<OPEN>','<CLOSE>','<HIGH>','<LOW>'])
gbpusd = pd.read_csv('GBPUSD_H1.csv', sep='\t', usecols=['<DATE>','<TIME>','<OPEN>','<CLOSE>','<HIGH>','<LOW>'])
usdjpy = pd.read_csv('USDJPY_H1.csv', sep='\t', usecols=['<DATE>','<TIME>','<OPEN>','<CLOSE>','<HIGH>','<LOW>'])
nq = pd.read_csv('US100.cash_H1.csv', sep='\t', usecols=['<DATE>','<TIME>','<OPEN>','<CLOSE>','<HIGH>','<LOW>'])
dax = pd.read_csv('GER40.cash_H1.csv', sep='\t', usecols=['<DATE>','<TIME>','<OPEN>','<CLOSE>','<HIGH>','<LOW>'])
xauusd = pd.read_csv('XAUUSD_H1.csv', sep='\t', usecols=['<DATE>','<TIME>','<OPEN>','<CLOSE>','<HIGH>','<LOW>'])
```

```python
eurusd['DATE'] = pd.to_datetime(eurusd['<DATE>'] + ' ' + eurusd['<TIME>'])
gbpusd['DATE'] = pd.to_datetime(gbpusd['<DATE>'] + ' ' + gbpusd['<TIME>'])
usdjpy['DATE'] = pd.to_datetime(usdjpy['<DATE>'] + ' ' + usdjpy['<TIME>'])
nq['DATE'] = pd.to_datetime(nq['<DATE>'] + ' ' + nq['<TIME>'])
dax['DATE'] = pd.to_datetime(dax['<DATE>'] + ' ' + dax['<TIME>'])
xauusd['DATE'] = pd.to_datetime(xauusd['<DATE>'] + ' ' + xauusd['<TIME>'])
```

```python
eurusd.set_index('DATE',inplace=True)
eurusd = eurusd.drop(['<DATE>','<TIME>'], axis=1)

gbpusd.set_index('DATE',inplace=True)
gbpusd = gbpusd.drop(['<DATE>','<TIME>'], axis=1)

usdjpy.set_index('DATE',inplace=True)
usdjpy = usdjpy.drop(['<DATE>','<TIME>'], axis=1)

nq.set_index('DATE',inplace=True)
nq = nq.drop(['<DATE>','<TIME>'], axis=1)

dax.set_index('DATE',inplace=True)
dax = dax.drop(['<DATE>','<TIME>'], axis=1)

xauusd.set_index('DATE',inplace=True)
xauusd = xauusd.drop(['<DATE>','<TIME>'], axis=1)
```

```python
#fuso horário UTC+3
eurusd.index = eurusd.index.tz_localize('Etc/GMT-3')
eurusd.index = eurusd.index.tz_convert('UTC')

gbpusd.index = gbpusd.index.tz_localize('Etc/GMT-3')
gbpusd.index = gbpusd.index.tz_convert('UTC')

usdjpy.index = usdjpy.index.tz_localize('Etc/GMT-3')
usdjpy.index = usdjpy.index.tz_convert('UTC')

nq.index = nq.index.tz_localize('Etc/GMT-3')
nq.index = nq.index.tz_convert('UTC')

dax.index = dax.index.tz_localize('Etc/GMT-3')
dax.index = dax.index.tz_convert('UTC')

xauusd.index = xauusd.index.tz_localize('Etc/GMT-3')
xauusd.index = xauusd.index.tz_convert('UTC')
```

```python
analise_eurusd = pd.DataFrame({
    'asia': eurusd_return_asia,
    'london': eurusd_return_london,
    'ny': eurusd_return_ny
}).dropna()


analise_gbpusd = pd.DataFrame({
    'asia': gbpusd_return_asia,
    'london': gbpusd_return_london,
    'ny': gbpusd_return_ny
}).dropna()


analise_usdjpy = pd.DataFrame({
    'asia': usdjpy_return_asia,
    'london': usdjpy_return_london,
    'ny': usdjpy_return_ny
}).dropna()


analise_nq = pd.DataFrame({
    'asia': nq_return_asia,
    'london': nq_return_london,
    'ny': nq_return_ny
}).dropna()


analise_dax = pd.DataFrame({
    'asia': dax_return_asia,
    'london': dax_return_london,
    'ny': dax_return_ny
}).dropna()


analise_xauusd = pd.DataFrame({
    'asia': xauusd_return_asia,
    'london': xauusd_return_london,
    'ny': xauusd_return_ny
}).dropna()

```

```python
analise_eurusd['sinal_asia'] = np.sign(analise_eurusd['asia'])
analise_eurusd['sinal_london'] = np.sign(analise_eurusd['london'])
analise_eurusd['sinal_ny'] = np.sign(analise_eurusd['ny'])

def classificar(row):
    if row['sinal_asia'] > 0 and row['sinal_london'] > 0:
        return 'asia e londres positivas'
    elif row['sinal_asia'] < 0 and row['sinal_london'] < 0:
        return 'asia e londres negativas'
    else:
        return 'divergentes'

analise_eurusd['cenario (eurusd)'] = analise_eurusd.apply(classificar, axis=1)

analise_eurusd['ny_reverteu'] = analise_eurusd['sinal_ny'] != analise_eurusd['sinal_asia']
resumo_eurusd = analise_eurusd.groupby('cenario (eurusd)').agg(
    total_dias=('ny_reverteu', 'count'),
    ny_reverteu=('ny_reverteu','sum'),
    taxa_reversao=('ny_reverteu','mean'),
    retorno_medio_ny=('ny','mean')
).round(4)

print(resumo_eurusd)

#Se asia e londres divergirem, para onde ny vai?
divergentes = analise_eurusd[analise_eurusd['cenario (eurusd)'] == 'divergentes'].copy()

divergentes['seguiu_asia'] = np.sign(divergentes['ny']) == np.sign(divergentes['asia'])
divergentes['seguiu_london'] = np.sign(divergentes['ny']) == np.sign(divergentes['london'])

print(f'\nDivergentes:')
print(f'Ny seguiu asia: {divergentes["seguiu_asia"].mean():.1%}')
print(f'Ny seguiu Londres: {divergentes["seguiu_london"].mean():.1%}')

#Dias em que asia e londres fizeram movimentos direcionais >=0.15% (mediana)
forte = analise_eurusd[
    (analise_eurusd['asia'].abs() >= 0.0015) &
    (analise_eurusd['london'].abs() >= 0.0015)
].copy()

resumo_forte = forte.groupby('cenario (eurusd)').agg(
    total_dias=('ny_reverteu','count'),
    ny_reverteu=('ny_reverteu','sum'),
    taxa_reversao=('ny_reverteu','mean'),
    retorno_mediop_ny=('ny','mean')
).round(4)

print(f'\nResumo com movimento forte >=0.15% (mediana):')
print(resumo_forte)

#asia e londres divergentes e fortes, ny segue quem?
divergente_forte = forte[forte['cenario (eurusd)'] == 'divergentes'].copy()

divergente_forte['seguiu_asia'] = np.sign(divergente_forte['ny']) == np.sign(divergente_forte['asia'])
divergente_forte['seguiu_london'] = np.sign(divergente_forte['ny']) == np.sign(divergente_forte['london'])

print(f'\ndivergente com força >=0.15% (mediana):')
print(f'Ny seguiu asia: {divergente_forte["seguiu_asia"].mean():.1%}')
print(f'Ny seguiu Londres: {divergente_forte["seguiu_london"].mean():.1%}')
print(f'Total de dias: {len(divergente_forte)}')
```
